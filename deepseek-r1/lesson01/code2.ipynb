## 我们的训练数据集

虽然本文没有具体说明强化学习预训练的初始数据集，但我们假设它应该以推理为重点。

为了尽可能接近原始复制结果，我们将使用以下两个开源推理 Hugging Face 数据集：

1. [NuminaMath-TIR](https://huggingface.co/datasets/AI-MO/NuminaMath-TIR)（用于 R1 零训练）

2. [Bespoke-Stratos-17k](https://huggingface.co/datasets/bespokelabs/Bespoke-Stratos-17k)（用于 R1 训练）

AI-MO/NuminaMath-TIR 包含 7 万道数学题，其信息栏展示了解决方案背后的 COT（思想链）推理。

| 字段 | 描述 |
|----------|------------|
| 问题 | 数学问题 |
| 解决方案 |分步解决方案 |
| 消息 | 聊天解决问题 |

查看示例：

```python
from datasets import load_dataset
# 从 DigitalLearningGmbH 加载“AI-MO/NuminaMath-TIR”数据集
MATH_le = load_dataset("AI-MO/NuminaMath-TIR", "default")  

# 访问训练集中的第一个样本
MATH_le['train'][0]
```

而 Bespoke-Stratos 包含 17K 个专注于数学和代码的问题。

```python
# 从 bespokelabs 加载“Bespoke-Stratos-17k”数据集
bespoke_rl = load_dataset("bespokelabs/Bespoke-Stratos-17k", "default")

# 访问训练集中的第一个样本
bespoke_rl['train'][0]
```

没有必要只选择这些数据集，你可以选择任何一个，只要它以推理为中心（**一个问题及其逐步解决方案**）。
